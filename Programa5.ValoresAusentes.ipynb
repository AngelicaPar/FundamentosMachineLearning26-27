**Imputación valores ausentes:** Es el proceso de reemplazar datos faltantes en un conjunto de datos como: registros con espacios en blanco, NaN o nulos por valores estimados o lógicos.

**SimpleImputer:** Permite completar valores faltantes en un conjunto de datos utilizando la media, mediana, moda o un valor constante.

```python
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
# Carga de datos
X_train = [[1, 4],
[np.nan, 3],
[7, 8]]
```

```python
# Limpieza de datos: imputación valores ausentes (modelo)
imputer = SimpleImputer(strategy='mean') # Utilizamos la estrategia de la media
# Aplicar los métodos "fit" y "transform" para imputar los valores pérdidos de X_train

X_train_clean = imputer.fit_transform(X_train
                                     )
print(X_train_clean)
```

```python
# Imputación de valores ausentes sobre el conjunto de test
X_test = [[np.nan, 10],
[6, np.nan],
[8, 2]]
```

```python
# Imputar los datos del conjunto de test
X_test_clean=imputer.transform(X_test)
print(X_test_clean)
```
