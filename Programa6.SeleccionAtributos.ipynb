**Selección de atributos:** Es una técnica utilizada en el aprendizaje automático y la ciencia de datos para elegir las variables más relevantes de un conjunto de datos antes de entrenar un modelo. Su objetivo principal es eliminar los datos redundantes o irrelevantes para mejorar el rendimiento y la eficiencia del sistema.

**VarianceThreshold:** Es un selector de características que elimina las variables de baja variación, es decir las que tienen una varianza menor o igual al umbral definido.

```python
import numpy as np
from sklearn.feature_selection import VarianceThreshold
```

```python
# Carga de datos.
X = [[0, 0, 1], [0, 1, 0], [1, 0, 0],
    [0, 1, 1], [0, 1, 0], [0, 1, 1]]
print(X)
```

```python
# Limpieza de datos: selección de atributos.
# fijamos un umbral (th) para decidir qué porcentaje mínimo deben variar los valores de una columna
# para considerarse una columna relevante a la hora de mantenerla en nuestro estudio.

th = 0.2
sel = VarianceThreshold(th)
print(sel.fit_transform(X))
```
