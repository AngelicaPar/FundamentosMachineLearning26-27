**Estandarización:** Proceso en el cual se transforma nuestro conjunto de datos para que cada característica tenga media 0 y desviación estándar 1. Para ello a cada valor se le resta la media de su característica y se divide entre su desviación estándar.

**Formula:**
$X_j^S = \frac{X_{ij} - \text{media}(X_j)}{\text{desv}(X_j)}$


```python
from sklearn import preprocessing
import numpy as np
```

```python
# Carga de datos.
X_train = np.array([[ 1., -1., 2.],
                    [ 2., 0., 0.],
                    [ 0., 1., -1.]])
print(X_train)
```


**StandardScaler**: Es una herramienta de preprocesamiento de datos que se usa para estandarizar las características numéricas (media 0 y desviación estándar 1). Se usa en algoritmos sensibles a la escala, como K-Nearest Neighbors, K-Means, Regresión Logística, SVM y Redes Neuronales, y funciona mejor cuando los datos siguen una distribución aproximadamente gaussiana.


```python
# Limpieza de datos: estandarización.
standardizer = preprocessing.StandardScaler()

# Estandarizamos los datos de entrenamiento
X_train_std = standardizer.fit_transform(X_train)
print(X_train_std)
```

```python
X_train_2 = np.array([[ 1., -1., 2.],
                      [ 2., 0., 0.],
                      [ 0., 1., -1.]])
```

```python
# Aplicamos los métodos "fit" y "transform" por separado
f = preprocessing.StandardScaler()
f = f.fit(X_train)
X_train_2_std = f.transform(X_train_2)
print(X_train_2_std)
```

```python
# Comprobar si las matrices estandarizadas son iguales
# ???
print(np.array_equiv(X_train_std, X_train_2_std))
```

```python
# Generar un conjunto de datos de test y estandarizar
# El conjunto de datos debe tener un tamaño 3*3
X_test = np.array([[ 3., -2., 0.],
                    [ -2., 5., 4.],
                    [ 1., -3., 2.]])
print(X_test)
```

```python
# Estandarizar los datos de "X_test" (SOLO APLICA EL MÉTODO "transform", NO EL "fit")

X_test_std = standardizer.transform(X_test)
print(X_test_std)

# Nota: En el conjunto de datos test se usa "transform", con la media y la desviación del entrenamiento,
# sin recalcularlas con el test es por eso que el resultado no tiene media 0 ni desviación 1.
```
