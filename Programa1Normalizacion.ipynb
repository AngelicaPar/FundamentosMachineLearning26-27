**Normalización:** Proceso en el cual convertimos nuestros datos en un rango de 0 y 1. A este proceso se lo denomina homotecia.

**Formula:**
$X_j^N = \frac{X_{ij} - \min(X_j)}{\max(X_j) - \min(X_j)}$

```python
from sklearn import preprocessing
import numpy as np
```

**Preprocessing**: Es un paquete de scikit-learn que permite utilizar muchas funciones para transformar distintas características de los datos como: normalización, estandarización o eliminación de la media y escalado de la varianza.

**MinMaxScaler**: Función que permite encontrar los valores mínimos y máximos de un conjunto de datos.

**numpy**: Una función de python desarrollada en C que me permite realizar operaciones matemáticas complejas con estructuras de datos como: matrices, vectores.


```python
# Carga de datos.
X_train = np.array([[ 1., -4.,  5.],
                    [ 5.,  0.,  2.],
                    [ 0.,  6., -3.]])

print(X_train)
```

```python
# Limpieza de datos: normalización.
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
```

**fit**: Calcula el mínimo y el máximo de cada columna de los datos de entrenamiento.

**transform**: Usa el mínimo y el máximo calculados por fit para convertir los datos en datos normalizados.

**fit_transform**: Hace los dos pasos juntos en uno solo.

```python
# Salida de resultados.
print(X_train_norm)
```

```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos "fit" y "transform" por separado
X_train_2 = np.copy(X_train)

# Crea un objeto "fitted_normalizer" llamando al método "fit"
f = normalizer.fit(X_train_2)

# Aplica el objeto sobre los datos "X_train_2" para obtener "X_train_norm_2"
X_train_norm_2 = f.transform(X_train_2)

# Imprimo la matriz X_train-norm_2
print(X_train_norm_2)

# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

```python
# Normalizar un nuevo conjunto de datos de test (SOLO HAY QUE APLICAR EL MÉTODO "transform", NO EL "fit")
X_test = np.array([[ 1., -3.,  2.],
                   [ 3.,  0.,  0.],
                   [ 0.,  4., -1.]])

# Normalizar los datos de "X_test"
X_test_norm = normalizer.transform(X_test)
print(X_test_norm)
```
